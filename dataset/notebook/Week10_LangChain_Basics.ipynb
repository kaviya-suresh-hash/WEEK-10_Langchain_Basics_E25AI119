```python
import pandas as pd
import os

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

print("LangChain libraries imported successfully!")
```

```python
MODEL_NAME = "gemma3:1b"

llm = ChatOllama(
    model=MODEL_NAME,
    temperature=0.3
)

print("Ollama model connected successfully!")
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an Education Assistant. Explain topics in simple English with examples."),
    ("human", "Topic: {topic}\nQuestion: {question}")
])

chain = prompt | llm

print("Prompt Template and LLM Chain created successfully!")
```

```python
topic = input("Enter the topic: ")
question = input("Enter your question: ")

response = chain.invoke({
    "topic": topic,
    "question": question
})

print("\nChatbot Response:")
print(response.content)
```

```python
import csv
import os
from langchain_core.prompts import ChatPromptTemplate

# Original Prompt
original_prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer the student's question."),
    ("human", "Topic: {topic}\nQuestion: {question}")
])

# Improved Prompt
improved_prompt = ChatPromptTemplate.from_messages([
    ("system", """You are an expert college tutor.
    Explain the answer in simple English.
    Organize the response using clear points.
    Include a suitable example wherever possible.
    Make the answer accurate, helpful and easy for students to understand."""),
    ("human", "Topic: {topic}\nQuestion: {question}")
])

# Test Questions
comparison_queries = [
    ("Python", "What is Python?"),
    ("Java", "Explain inheritance in Java."),
    ("DBMS", "What is a primary key?"),
    ("Operating System", "What is an operating system?"),
    ("Artificial Intelligence", "Explain artificial intelligence.")
]

# Create Comparison File
os.makedirs("dataset", exist_ok=True)

with open("dataset/prompt_comparison.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)
    writer.writerow([
        "Topic",
        "Question",
        "Original_Response",
        "Improved_Response",
        "Observation"
    ])

    for topic, question in comparison_queries:

        print("\nTopic:", topic)
        print("Question:", question)

        # Original Response
        original_chain = original_prompt | llm
        original_response = original_chain.invoke({
            "topic": topic,
            "question": question
        })

        # Improved Response
        improved_chain = improved_prompt | llm
        improved_response = improved_chain.invoke({
            "topic": topic,
            "question": question
        })

        print("\nOriginal Response:")
        print(original_response.content)

        print("\nImproved Response:")
        print(improved_response.content)

        writer.writerow([
            topic,
            question,
            original_response.content,
            improved_response.content,
            "Compare clarity, examples and organization"
        ])

print("\nOriginal and Improved Prompt Comparison Completed!")
```

```python
import os
import csv
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

# Model Configuration
llm = ChatOllama(
    model="gemma3:1b",
    temperature=0.3
)

# Create Dataset Folder
os.makedirs("dataset", exist_ok=True)

# -------------------------------
# 1. Summarization Chain
# -------------------------------

summary_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a summarization assistant. Summarize the given text in simple and clear English."),
    ("human", "Summarize this text:\n{text}")
])

summarization_chain = summary_prompt | llm

sample_text = """
Artificial Intelligence is a technology that enables machines
to perform tasks that normally require human intelligence.
It is used in healthcare, education, banking and transportation.
AI helps improve efficiency, automate tasks and support decision-making.
"""

summary_result = summarization_chain.invoke({
    "text": sample_text
})

print("SUMMARIZATION CHAIN")
print(summary_result.content)


# -------------------------------
# 2. Question Answering Chain
# -------------------------------

qa_prompt = ChatPromptTemplate.from_messages([
    ("system", "Answer the question using the given context. Use simple English."),
    ("human", "Context: {context}\nQuestion: {question}")
])

qa_chain = qa_prompt | llm

context = """
Python is a high-level programming language.
It is easy to learn and has a simple syntax.
Python is used in web development, data science and artificial intelligence.
"""

qa_result = qa_chain.invoke({
    "context": context,
    "question": "What are the uses of Python?"
})

print("\nQUESTION ANSWERING CHAIN")
print(qa_result.content)


# -------------------------------
# Save Chain Results
# -------------------------------

with open("dataset/additional_chains_results.csv",
          "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow(["Chain_Name", "Input", "Output"])

    writer.writerow([
        "Summarization",
        sample_text,
        summary_result.content
    ])

    writer.writerow([
        "Question Answering",
        context,
        qa_result.content
    ])

print("\nBoth Additional Chains Executed Successfully!")
print("Results saved in CSV file!")
```
